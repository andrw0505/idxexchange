```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
sold_dfs = []

for year in [2024, 2026]:
    for month in range(1, 13):
        fname = f"CRMLSSold{year}{month:02d}.csv"
        try:
            df = pd.read_csv(fname)
            df['Year'] = year
            df['Month'] = month
            sold_dfs .append(df)
        except FileNotFoundError:
            # Skip months that don't exist yet
            continue

# If you want one big DataFrame:
all_sold_df = pd.concat(sold_dfs, ignore_index=True)
```
