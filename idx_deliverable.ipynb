```python
# Making a unified listing df
listing_dfs = []

for year in [2024, 2026]:
    for month in range(1, 13):
        fname = f"CRMLSListing{year}{month:02d}.csv"
        try:
            df = pd.read_csv(fname,parse_dates=['CloseDate'])
            df['Year'] = year
            df['Month'] = month
            listing_dfs.append(df)
        except FileNotFoundError:
            # Skip months that don't exist yet
            continue

# If you want one big DataFrame:
all_listings_df = pd.concat(listing_dfs, ignore_index=True)
```

```python
# Re-read 'enriched_sold.csv' with correct date format
enriched_sold = pd.read_csv('enriched_sold.csv',index_col=0,parse_dates=['CloseDate','PurchaseContractDate','ListingContractDate','ContractStatusChangeDate'])
enriched_sold.info()
```

```python
# Re-read 'enriched_listings.csv' with correct date format
enriched_listing = pd.read_csv('enriched_listing.csv',index_col=0,parse_dates=['CloseDate','PurchaseContractDate','ListingContractDate','ContractStatusChangeDate'])
enriched_listing.info()
```

```python
# Checking if date follows the correct order

# Check if listing is after close

def list_after_close(listing_date,close_date):
    return listing_date > close_date # True = faulty value, False = correct value

def purchase_after_close(purchase_date,close_date):
    return purchase_date > close_date # True = faulty value, False = correct value


enriched_sold['list_after_close_flag']= enriched_sold.apply(
    lambda row: list_after_close(row['ListingContractDate'],row['CloseDate']),
                axis=1
)

enriched_sold['purchase_after_close_flag']=enriched_sold.apply(
    lambda row: purchase_after_close(row['PurchaseContractDate'],row['CloseDate']),
    axis=1
)
```

```python
enriched_sold['negative_timeline_flag'] = (
    enriched_sold['list_after_close_flag'] | enriched_sold['purchase_after_close_flag']
)

enriched_sold_postimeline = enriched_sold[enriched_sold['negative_timeline_flag']==False].copy()
print(enriched_sold_postimeline.info())
display(enriched_sold_postimeline.head())
```

```python
enriched_listing['list_after_close_flag']= enriched_listing.apply(
    lambda row: list_after_close(row['ListingContractDate'],row['CloseDate']),
                axis=1
)

enriched_listing['purchase_after_close_flag']=enriched_listing.apply(
    lambda row: purchase_after_close(row['PurchaseContractDate'],row['CloseDate']),
    axis=1
)
```

```python
enriched_listing['negative_timeline_flag'] = (
    enriched_listing['list_after_close_flag'] | enriched_listing['purchase_after_close_flag']
)
enriched_listing_postimeline = enriched_listing[enriched_listing['negative_timeline_flag']==False].copy()
print(enriched_listing_postimeline.info())
display(enriched_listing_postimeline.head())
```

```python
# Checking Longtitude & Latitude for valid format
enriched_listing_postimeline['Latitude']=pd.to_numeric(enriched_listing_postimeline['Latitude'],errors='coerce')
enriched_listing_postimeline['Longitude']=pd.to_numeric(enriched_listing_postimeline['Longitude'],errors='coerce')

enriched_sold_postimeline['Latitude']=pd.to_numeric(enriched_sold_postimeline['Latitude'],errors='coerce')
enriched_sold_postimeline['Longitude']=pd.to_numeric(enriched_sold_postimeline['Longitude'],errors='coerce')

def check_coord(longitude,latitude):
    if not pd.isna(longitude) and not pd.isna(latitude):
        if (longitude < 0) and (latitude != 0):
            if (32.5 <= latitude <= 42.0) and (-124.5 <= longitude <= -114.0):
                return True # True = coordinate is valid and pass all checks
    return False
            
enriched_listing_postimeline['correct_coord'] = enriched_listing_postimeline.apply(
    lambda row: check_coord(row['Longitude'],row['Latitude']),
    axis=1
)

enriched_sold_postimeline['correct_coord'] = enriched_sold_postimeline.apply(
    lambda row: check_coord(row['Longitude'],row['Latitude']),
    axis=1
)
```

```python
print('Correct Coordinates Value Counts for Sold Data\n')
print(enriched_sold_postimeline['correct_coord'].value_counts())
print('\nCorrect Coordinates Value Counts for Listing Data\n')
print(enriched_listing_postimeline['correct_coord'].value_counts())
```

```python
enriched_sold_v3 = enriched_sold_postimeline[enriched_sold_postimeline['correct_coord']==True]
print(enriched_sold_v3.info())
display(enriched_sold_v3.head())
```

```python
enriched_listing_v3 = enriched_listing_postimeline[enriched_listing_postimeline['correct_coord']==True]
print(enriched_listing_v3.info())
display(enriched_listing_v3.head())
```

```python
# Find the difference in rows after checking for date and coordinates

sold_r_change = len(enriched_sold)-len(enriched_sold_v3)

listing_r_change = len(enriched_listing)-len(enriched_listing_v3)

print(f'After checking for correct timeline and coordinates, {sold_r_change} rows are removed for Sold data')
print(f'After checking for correct timeline and coordinates, {listing_r_change} rows are removed for Listing data')
```

# Week 7- Outlier Detection and Data Quality

```python
def flagging_outlier(df,column):
    Q1 = df[column].quantile(0.25)
    Q3 = df[column].quantile(0.75)
    IQR = Q3 - Q1
    lower = Q1 - 1.5 * IQR
    upper = Q3 + 1.5 * IQR

    return (df[column] < lower) | (df[column] > upper)
```

```python
enriched_sold_v5 = enriched_sold_v4.copy()
enriched_sold_v5 = enriched_sold_v5[
    (enriched_sold_v5['ClosePrice'] > 0) &
    (enriched_sold_v5['LivingArea'] > 0) &
    (enriched_sold_v5['DaysOnMarket'] > 0)
]
```

```python
enriched_listing_v5 = enriched_listing_v4.copy()
enriched_listing_v5 = enriched_listing_v5[
    (enriched_listing_v5['ClosePrice'] > 0) &
    (enriched_listing_v5['LivingArea'] > 0) &
    (enriched_listing_v5['DaysOnMarket'] > 0)
]
```

```python
enriched_sold_v6 = enriched_sold_v5.copy()
enriched_sold_v6['price_outlier']=flagging_outlier(enriched_sold_v5,'ClosePrice')
enriched_sold_v6['area_outlier']=flagging_outlier(enriched_sold_v5,'LivingArea')
enriched_sold_v6['daysonmarket_outlier']=flagging_outlier(enriched_sold_v5,'DaysOnMarket')
```

```python
enriched_listing_v6 = enriched_sold_v5.copy()
enriched_listing_v6['price_outlier']=flagging_outlier(enriched_listing_v6,'ClosePrice')
enriched_listing_v6['area_outlier']=flagging_outlier(enriched_listing_v6,'LivingArea')
enriched_listing_v6['daysonmarket_outlier']=flagging_outlier(enriched_listing_v6,'DaysOnMarket')
```
