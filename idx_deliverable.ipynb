```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

url = "https://fred.stlouisfed.org/graph/fredgraph.csv?id=MORTGAGE30US"

mortgage = pd.read_csv(url, parse_dates=['observation_date'])

mortgage.columns = ['date', 'rate_30yr_fixed']
```

```python
# Making a unified listing df
listing_dfs = []

for year in [2024, 2026]:
    for month in range(1, 13):
        fname = f"CRMLSListing{year}{month:02d}.csv"
        try:
            df = pd.read_csv(fname,parse_dates=['CloseDate'])
            df['Year'] = year
            df['Month'] = month
            listing_dfs.append(df)
        except FileNotFoundError:
            # Skip months that don't exist yet
            continue

# If you want one big DataFrame:
all_listings_df = pd.concat(listing_dfs, ignore_index=True)
```

```python
all_listings_df.info()
```

```python
print("Preview of Concatenated Listing DataFrame")
display(all_listings_df.tail())

print("---------/--------")
print("All columns in DataFrame\n")
print(all_listings_df.columns)
print("---------/--------")
print("Description of DataFrame\n")
print(all_listings_df.describe)
```

```python
cleaned_listings=all_listings_df[all_listings_df['PropertyType']=='Residential']
cleaned_listings.info()
display(cleaned_listings.head())
```

```python
before_r_list = all_listings_df.shape[0]
after_r_list = cleaned_listings.shape[0]

print(f"Row count before Residential filter: {before_r_list} rows")
print(f"Row count after Residential filter: {after_r_list} rows")
row_change = before_r_list - after_r_list
print(f"{row_change} rows were dropped after filtering for 'Residential' property only.")
```

```python
cleaned_listings.to_csv("cleaned_listings.csv",index=True,encoding="utf-8-sig")
```

```python
sold_dfs = []

for year in [2024, 2026]:
    for month in range(1, 13):
        fname = f"CRMLSSold{year}{month:02d}.csv"
        try:
            df = pd.read_csv(fname)
            df['Year'] = year
            df['Month'] = month
            sold_dfs.append(df)
        except FileNotFoundError:
            # Skip months that don't exist yet
            continue

# If you want one big DataFrame:
all_sold_df = pd.concat(sold_dfs, ignore_index=True)
```

```python
all_sold_df.info()
```

```python
print("Preview of Concatenated Sold DataFrame")
display(all_sold_df.head())

print("---------/--------")
print("All columns in DataFrame\n")
print(all_sold_df.columns)
print("---------/--------")
print("Description of DataFrame\n")
print(all_sold_df.describe)

before_shape = all_sold_df.shape
```

```python
prop_type = all_sold_df['PropertyType'].unique()
print(f"Unique Property Types: {prop_type}")
```

```python
residential_sold_df=all_sold_df[all_sold_df['PropertyType']=='Residential']
residential_sold_df.info()
display(residential_sold_df.tail())

after_shape = residential_sold_df.shape
```

```python
# Change in rows
print(f"Row count before Residential filter: {before_shape[0]} rows")
print(f"Row count after Residential filter: {after_shape[0]} rows")
row_change = before_shape[0] - after_shape[0]
print(f"{row_change} rows were dropped after filtering for 'Residential' property only.")
```

```python
completedness_count = pd.DataFrame(
    residential_sold_df.isnull().sum()
).reset_index()
completedness_count.columns = ["Column Name", "count"]
completedness_count = completedness_count.sort_values(by='count',ascending=False)

display(completedness_count)

completedness_pct = pd.DataFrame(residential_sold_df.isnull().mean()).reset_index()
completedness_pct.columns = ["Column Name","pct"]
```

```python
flagged_columns = completedness_pct[completedness_pct['pct']>=0.9].sort_values(by='pct',ascending=False).reset_index(drop=True)
display(flagged_columns)

lst_flagged = flagged_columns["Column Name"].to_list()
print(lst_flagged)
```

```python
cleaned_sold = all_sold_df.drop(columns=lst_flagged)

print(f"Number of Columns before filtering out flagged columns: {len(all_sold_df.columns.to_list())} columns.")
print(f"Number of Columns after filtering out flagged columns: {len(cleaned_sold.columns.to_list())} columns.")
print(f"{len(lst_flagged)} columns are removed due to high percentage of missing values.")
```

```python
#Summary table for ClosePrice, LivingArea, and DaysOnMarket
cols = ["ClosePrice","LivingArea","DaysOnMarket"]

summary = cleaned_sold[cols].agg([
    "min","max","mean","median",
    lambda x: x.quantile(0.25),
    lambda x: x.quantile(0.75),
    lambda x: x.quantile(0.90),
    lambda x: x.quantile(0.95)
])

summary.index = ["min", "max", "mean", "median", "p25", "p75", "p90", "p95"]

display(summary.T)
```

```python
cleaned_sold.to_csv("cleaned_sold.csv",index=True,encoding="utf-8-sig")
```

```python
# Combining FRED data
mortgage

#Resample

mortgage['year_month'] = mortgage['date'].dt.to_period('M')

monthly_mortgage = mortgage.groupby('year_month')["rate_30yr_fixed"].mean().reset_index()

display(monthly_mortgage)
```

```python
#Re-format date columns for Sold and Listing DFs

cleaned_sold["year_month"] = pd.to_datetime(cleaned_sold["CloseDate"]).dt.to_period("M")

cleaned_listings['year_month'] = pd.to_datetime(cleaned_listings['ListingContractDate']).dt.to_period('M')
```

```python
sold_w_rates = cleaned_sold.merge(monthly_mortgage,on="year_month",how="left")

listing_w_rates = cleaned_listings.merge(monthly_mortgage,on="year_month",how="left")
```

```python
print(sold_w_rates['rate_30yr_fixed'].isnull().sum())
print(listing_w_rates['rate_30yr_fixed'].isnull().sum())

display(sold_w_rates[['CloseDate','year_month','ClosePrice','rate_30yr_fixed']].head())
```

```python
sold_w_rates.to_csv('enriched_sold.csv',index=True,encoding='utf-8-sig')
listing_w_rates.to_csv('enriched_listing.csv',index=True,encoding='utf-8-sig')
```
